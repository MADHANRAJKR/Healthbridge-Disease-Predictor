# HealthBridge — 01: Exploratory Data Analysis

This notebook explores the supplied dataset and the augmented 30-disease dataset.


```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style='darkgrid', palette='husl')
ROOT = Path('..')
```

## 1. Load supplied data

```python
df_raw = pd.read_csv(ROOT / 'data/raw/aiml_healthcare_symptoms.csv')
print(f'Shape: {df_raw.shape}')
df_raw.head()
```

## 2. Missing values & duplicates

```python
print('Missing values:'); print(df_raw.isnull().sum())
print(f'\nDuplicate rows: {df_raw.duplicated().sum()}')
```

## 3. Class distribution (supplied 8 classes)

```python
fig, ax = plt.subplots(figsize=(10, 4))
df_raw['Diagnosis'].value_counts().plot(kind='bar', ax=ax, color=sns.color_palette('husl', 8))
ax.set_title('Class Distribution — Supplied Dataset (8 diseases)')
ax.set_ylabel('Count'); ax.set_xlabel('')
plt.xticks(rotation=30, ha='right'); plt.tight_layout(); plt.show()
```

## 4. Feature statistics

```python
df_raw.describe().round(2)
```

## 5. Correlation heatmap

```python
num_cols = df_raw.select_dtypes(include='number').columns.drop('Patient_ID', errors='ignore')
fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(df_raw[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', ax=ax, linewidths=0.5)
ax.set_title('Feature Correlation Heatmap')
plt.tight_layout(); plt.show()
```

## 6. Load augmented 30-disease dataset

```python
X = pd.read_csv(ROOT / 'data/processed/features.csv')
y = pd.read_csv(ROOT / 'data/processed/labels.csv')['Diagnosis']
print(f'Total rows: {len(X)}  |  Disease classes: {y.nunique()}')
print('Classes:', sorted(y.unique().tolist()))
```

## 7. Class distribution (30 diseases)

```python
fig, ax = plt.subplots(figsize=(16, 5))
y.value_counts().sort_index().plot(kind='bar', ax=ax, color=sns.color_palette('husl', y.nunique()))
ax.set_title('Class Distribution — Augmented Dataset (30 diseases)')
ax.set_ylabel('Count'); ax.set_xlabel('')
plt.xticks(rotation=45, ha='right'); plt.tight_layout(); plt.show()
```

## 8. Symptom prevalence by disease (binary features)

```python
binary_feats = ['Fever','Cough','Fatigue','Breathlessness','Chest_Pain','Nausea','Headache']
df_all = X.copy(); df_all['Diagnosis'] = y.values
pivot = df_all.groupby('Diagnosis')[binary_feats].mean()
fig, ax = plt.subplots(figsize=(14, 10))
sns.heatmap(pivot, annot=True, fmt='.2f', cmap='YlOrRd', ax=ax, linewidths=0.3)
ax.set_title('Symptom Prevalence by Disease (mean binary rate)')
plt.tight_layout(); plt.show()
```
